### Tic Tac Toe

```python
def display_board(board):
    """
       This function is designed to display a user friendly board for visualisation of the game state

       The parameter is the current board which will be a 2D list/matrix of which tile each player has used
       I made this function as the board will be displayed after every move, hence creatign a function would reduce clutter
    """
    
    print(board[0][0], "|", board[0][1], "|", board[0][2])
    print("--+---+--")
    print(board[1][0], "|", board[1][1], "|", board[1][2])
    print("--+---+--")
    print(board[2][0], "|", board[2][1], "|", board[2][2])


def check_winner(player_moves):
     """
    This functions checks to see if a player has won the game. 

    It's parameter is the list of moves each player has. 

    It works by checking if all the values in any of the winning combinations are within the list of a players moves.
    This approach works well for the small grid of tic tac toe, but doesn't scale well as the board gets larger.
    """
    winning_combinations = [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9],
        [1, 4, 7],
        [2, 5, 8],
        [3, 6, 9],
        [1, 5, 9],
        [3, 5, 7]
    ]

    for combination in winning_combinations:
        if all(position in player_moves for position in combination):
            return True

    return False


def play_game():
    # Create an empty board
    board = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "]
    ]

    # Store the positions selected by each player
    player_x_moves = []
    player_o_moves = []

    # Store all positions that have already been used
    used_positions = []

    # True = Player 1's turn
    # False = Player 2's turn
    player_one = True

    # A maximum of 9 turns can be played
    for turn in range(9):

        display_board(board)

        # Ask the current player for a position
        while True:

            if player_one:
                user_input = input(
                    "Player 1 (X), choose a position (1-9): "
                )
            else:
                user_input = input(
                    "Player 2 (O), choose a position (1-9): "
                )

            # Try to convert the input into an integer
            try:
                position = int(user_input)

            except ValueError:
                print("Please enter a number between 1 and 9.")
                continue

            # Check that the position is between 1 and 9
            if position < 1 or position > 9:
                print("Please choose a position between 1 and 9.")

            # Check that the position has not already been used
            elif position in used_positions:
                print("That position has already been taken.")

            else:
                break

        # Convert the numpad position into a board row and column
        row = 2 - ((position - 1) // 3)
        column = (position - 1) % 3

        # Record the position
        used_positions.append(position)

        # Place the player's symbol on the board
        if player_one:
            board[row][column] = "X"
            player_x_moves.append(position)

            # Check whether Player 1 has won
            if check_winner(player_x_moves):
                display_board(board)
                print("Congratulations Player 1! You win!")
                return "Player 1"

        else:
            board[row][column] = "O"
            player_o_moves.append(position)

            # Check whether Player 2 has won
            if check_winner(player_o_moves):
                display_board(board)
                print("Congratulations Player 2! You win!")
                return "Player 2"

        # Check whether all nine positions have been used
        if len(used_positions) == 9:
            display_board(board)
            print("The game is a draw!")
            return "Draw"

        # Change to the other player
        player_one = not player_one


# Best of three

# Here I store the statistics of each game and return them to the player to see the overall match results

player_1_wins = 0
player_2_wins = 0
draws = 0

game_number = 1

while player_1_wins < 2 and player_2_wins < 2:

    print()
    print("====================")
    print("Game", game_number)
    print("====================")

    result = play_game()

    # Update the statistics
    if result == "Player 1":
        player_1_wins += 1

    elif result == "Player 2":
        player_2_wins += 1

    else:
        draws += 1

    game_number += 1

    print()
    print("Current statistics:")
    print("Player 1 wins:", player_1_wins)
    print("Player 2 wins:", player_2_wins)
    print("Draws:", draws)



# Final match result

print()
print("====================")
print("FINAL RESULTS")
print("====================")

print("Player 1 wins:", player_1_wins)
print("Player 2 wins:", player_2_wins)
print("Draws:", draws)

if player_1_wins == 2:
    print("Player 1 wins the match!")

elif player_2_wins == 2:
    print("Player 2 wins the match!")
```

### Hangman

```python
import random

word_banks = {
    "Animals": ["dog", "cat", "rabbit", "hamster", "horse"],
    "Food": ["pizza", "burger", "pasta", "sandwich", "chocolate"],
    "Countries": ["england", "france", "germany", "spain", "italy"]
}

hints = {
    "Animals": "The word is a type of pet",
    "Food": "The word is something you can eat",
    "Countries": "The word is a country"
}

category = random.choice(list(word_banks.keys()))
word = random.choice(word_banks[category])

hidden_word = ["_"] * len(word)

guessed_letters = []
wrong_guesses = 0
max_guesses = 6


print(" ".join(hidden_word))

while wrong_guesses < max_guesses:

    while True:
        guess = input("Guess a letter: ").lower()

        if len(guess) == 1 and guess.isalpha():
            break
        else:
            print("Please enter a single letter.")

    if guess in guessed_letters:
        print("You have already guessed that letter!")

    else:
        guessed_letters.append(guess)

        if guess in word:
            print("Correct!")
            print("")

            for i in range(len(word)):
                if word[i] == guess:
                    hidden_word[i] = guess

        else:
            wrong_guesses += 1
            print("Incorrect!")
            print("Guesses remaining:", max_guesses - wrong_guesses)
            print("")

            if max_guesses - wrong_guesses == 2:
                print("")
                print("Hint:", hints[category])
                print("")
                
    print(" ".join(hidden_word))

    # Check if the player has won
    if "_" not in hidden_word:
        print("Congratulations! The word was:", word)
        break

    # Check if the player has lost
    if wrong_guesses == max_guesses:
        print("You lost! The word was:", word)
        break
```
